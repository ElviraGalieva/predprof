```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```
import os

def find_file(filename, search_path='/'):
    """Поиск файла по имени"""
    for root, dirs, files in os.walk(search_path):
        if filename in files:
            print(f"Найден: {os.path.join(root, filename)}")

find_file('netflix_titles.csv')
```

```
netflix_overall = pd.read_csv("/usr/bin/netflix_titles.csv")
netflix_overall
```

```
# Выбор только столбца 'date_added' из DataFrame с сериалами, удаление пропусков
netflix_date = netflix_shows[['date_added']].dropna()

# Проверка, что пропусков больше нет
netflix_date.isna().sum()
```

```
# Просмотр первых строк обработанных данных
netflix_date.head()
```

```
# Извлечение года из даты добавления
netflix_date['year'] = netflix_date['date_added'].apply(lambda x : x.split(', ')[1])

# Извлечение месяца из даты добавления
# lstrip() удаляет пробелы слева, split(' ')[0] берет первое слово (название месяца)
netflix_date['month'] = netflix_date['date_added'].apply(lambda x : x.lstrip().split(' ')[0])
netflix_date.head()
```

```
# Сортировка данных по году и месяцу
netflix_date.sort_values(by=['year', 'month'])
```

```
# Создание порядка месяцев (от декабря к январю из-за [::-1] - обратный порядок)
month_order = ['January', 'February', 'March', 'April', 'May', 'June',
                'July', 'August', 'September', 'October', 'November', 'December'][::-1]
# Группировка по годам и месяцам, подсчет количества, создание таблицы
df = netflix_date.groupby('year')['month'].value_counts().unstack().fillna(0)[month_order].T
df
```

```
# Заполнение пропущенных значений в столбце 'country' на 'Unknown'
netflix_movies['country'].fillna('Unknown', inplace=True)

# Проверка, что пропусков больше нет
netflix_movies.isna().sum()
```

```
# Подсчет количества фильмов по странам (простой способ)
netflix_movies['country'].value_counts()
```

```
# Словарь для хранения количества фильмов по странам
countries_mov = {}
# Получение списка всех стран из столбца 'country'
cou = list(netflix_movies['country'])

# Проход по каждому значению страны
for i in cou:
    # Удаление пробелов (ОШИБКА: replace удаляет все пробелы, делая строку слитной)
    i = i.replace(' ', '')

    i = list(i.split(','))

    # Проход по каждой стране в строке (если несколько стран через запятую)
    for j in i:
        # Если страна уже есть в словаре, увеличиваем счетчик
        if j in list(countries_mov.keys()):
            countries_mov[j] += 1
        else:
            # Если нет - добавляем со значением 1
            countries_mov[j] = 1
```

```
# Вывод словаря со странами
countries_mov
```

```
# Сортировка словаря countries_mov по убыванию количества фильмов
# .items() возвращает пары (страна, количество)
# lambda item: countries_mov[item[0]] - функция для получения значения по ключу, по количеству фильмов в странах
countries_fin = sorted(countries_mov.items(), key=lambda item: countries_mov[item[0]], reverse=True)

# Оставляем только первые 10 (топ-10 стран)
countries_fin = countries_fin[:10]

# Вывод топ-10 стран с количеством фильмов
countries_fin
```

```
# Импорт Counter для подсчета частоты элементов
from collections import Counter

# Получение списка всех жанров из столбца 'listed_in'
genres = list(netflix_movies['listed_in'])

# Список для хранения отдельных жанров
gen = []

# Проход по каждой строке с жанрами
for i in genres:
    # ОШИБКА: split('.') - неправильный разделитель!
    # В данных жанры разделены запятыми, а не точками
    i = list(i.split('.'))

    # Проход по каждому жанру в строке
    for j in i:
        # Добавление жанра в общий список (с удалением точек)
        gen.append(j.replace('.', ''))

# Подсчет количества каждого жанра
g = Counter(gen)
```

```
# Создание TF-IDF векторизатора
# stop_words='english' - удаляет стандартные английские стоп-слова (the, and, of и т.д.)
tfidf = TfidfVectorizer(stop_words='english')

# Преобразование текстовых описаний в матрицу TF-IDF признаков
# fit_transform() - сначала обучается на данных, затем преобразует
tfidf_matrix = tfidf.fit_transform(netflix_movies['description'])

# Размер полученной матрицы
tfidf_matrix.shape
# Результат: (5377, 14601)
# 5377 - количество фильмов (строк)
# 14601 - количество уникальных слов (признаков)
```

```
from sklearn.metrics.pairwise import cosine_similarity

# Вычисление матрицы косинусного сходства между всеми фильмами
# Результат: квадратная матрица 5377x5377, где [i,j] - сходство между фильмом i и j
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)
cosine_sim
```

```
# Просмотр размера матрицы сходства (должно быть количество фильмов x количество фильмов)
cosine_sim.shape
```

```
# Создание Series для быстрого поиска индекса фильма по его названию
# drop_duplicates() - на случай, если есть фильмы с одинаковыми названиями
indices = pd.Series(netflix_movies.index, index=netflix_movies['title']).drop_duplicates()
indices
```

```
# Преобразование матрицы сходства в DataFrame для удобной работы
cosine_sim_df = pd.DataFrame(cosine_sim)

# Назначение названий фильмов в качестве названий столбцов
cosine_sim_df.columns = indices.index

# Добавление столбца с названиями и установка его как индекс
cosine_sim_df['title'] = indices.index
cosine_sim_df = cosine_sim_df.set_index('title')

# Просмотр первых строк
cosine_sim_df.head()
```

```
def get_recommendations(title, cosine_sim=cosine_sim_df):
    """
    Возвращает 10 фильмов, похожих на заданный фильм по описанию.

    Параметры:
        title: название фильма
        cosine_sim: DataFrame с матрицей косинусного сходства

    Возвращает:
        Series с названиями 10 наиболее похожих фильмов
    """
    # Получаем индекс фильма по названию
    idx = indices[title]

    # Получаем ряд сходства для этого фильма
    # enumerate превращает в список пар (индекс, значение)
    sim_scores = list(enumerate(cosine_sim_df.loc[title]))

    # Сортируем по убыванию сходства (самые похожие первые)
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # Берём первые 10, исключая сам фильм (индекс 0)
    sim_scores = sim_scores[1:11]

    # Извлекаем индексы фильмов
    movie_indices = [i[0] for i in sim_scores]

    # Возвращаем названия
    return netflix_movies['title'].iloc[movie_indices]
```

```
# Рекомендации, основанные на сюжете (описании фильмов)

# Поиск описания фильма 'Last Breath'
# .iloc[0] - берёт первое (и единственное) описание
netflix_movies[netflix_movies['title'] == 'Last Breath'].description.iloc[0]

```

```
# Получение рекомендаций для фильма 'Last Breath'
# Вызывает ранее созданную функцию get_recommendations()
get_recommendations('Last Breath')
```
